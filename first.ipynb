```python
#Create an API client
import anthropic
client = anthropic.Client()
model = "claude-sonnet-4-6"
```

```python
# Make a single request
message = client.messages.create(
    model = model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
```

```python
message.content[0].text
```

```python
# Make multi turn requests
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })
    return messages

def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })
    return messages

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text
```

```python
# Start with empty messages list
messages = []

# Add initial user message
add_user_message(messages, "What is quantum computing? Answer in one sentence")

# Get Claude's response
response = chat(messages)

# Add Claude's response to conversation history 
add_assistant_message(messages, response)

# Add follow up user message
add_user_message(messages, "Write Another sentence")

# Get full follow up response from Claude
final_answer = chat(messages)
```

```python
# Creating a chat system

# Start with empty messages list
messages = []

# Use a 'while true' loop to keep the conversation
while True:
    # Get user input
    user_input = input("You: ")
    print(f"You: {user_input}")

    # Add user message to conversation history
    add_user_message(messages, user_input)

    # Get Claude's response
    response = chat(messages)

    # Add Claude's response to conversation history
    add_assistant_message(messages, response)

    # Print Claude's response
    print(f"Claude: {response}")
```
