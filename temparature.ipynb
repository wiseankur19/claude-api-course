```python
# Load env variables from .env file
from dotenv import load_dotenv
load_dotenv()

#Create an API client
import anthropic
client = anthropic.Client()
model = "claude-haiku-4-5"
```

```python
# Helper functions to add messages to the conversation
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })
    return messages

def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })
    return messages

    

def chat(messages, temperature=1.0):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "extra_body": {
            "temperature": temperature
        }
    }


    message = client.messages.create(**params)

    return message.content[0].text
```

```python
messages = []

add_user_message(
    messages, 
    "Generate one sentence movie idea"
)

#answer = chat(messages, temperature=0.0) #It will give the same answer every time you run it

answer = chat(messages, temperature=1.0) #It will give different answers every time you run it

answer
```
