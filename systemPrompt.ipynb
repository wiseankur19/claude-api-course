```python
# Load env variables from .env file
from dotenv import load_dotenv
load_dotenv()

#Create an API client
import anthropic
client = anthropic.Client()
model = "claude-sonnet-4-6"
```

```python
# Make multi turn requests
def add_user_message(messages, content):
    messages.append({
        "role": "user",
        "content": content
    })
    return messages

def add_assistant_message(messages, content):
    messages.append({
        "role": "assistant",
        "content": content
    })
    return messages

def chat(messages):
    system = """
    You are a patient math tutor.
    Do not directly answer a student's questions.
    Guide them to a solution step by step.
    """
    message = client.messages.create(
        model = model,
        max_tokens=1000,
        messages=messages,
        system=system
    )
    return message.content[0].text

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens=1000,
        messages=messages
    )
    return message.content[0].text

def chatWithConstantSystemPrompt(messages):
    system = """
    You are a patient math tutor.
    Do not directly answer a student's questions.
    Guide them to a solution step by step.
    """
    message = client.messages.create(
        model = model,
        max_tokens=1000,
        messages=messages,
        system=system
    )
    return message.content[0].text

def chatWithSystemPromptParam(messages, system=None):
    
    message = client.messages.create(
        model = model,
        max_tokens=1000,
        messages=messages,
        system=system
    )
    return message.content[0].text


def chatWithSystemPromptParamHandling(messages, system=None):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }

    if(system is not None):
        params["system"] = system

    message = client.messages.create(**params)

    return message.content[0].text
```
